```python
import numpy as np
```

```python
def get_veivlet(len: int) -> np.array:
    return np.concatenate((np.ones(len // 2), -np.ones(len // 2)))
```

```python
def convolve_vk(vk: list[float], wavelet: list[float]) -> list:
    result = [0] * len(vk)
    for i in range(len(vk)):
        for j in range(len(wavelet)):
            if i - j >= 0 and i - j < len(vk):
                result[i] += (vk[i - j] * wavelet[j])
    return result
```

```python
first_vk = [
        complex(5, 0),
        complex(0, -4),
        complex(-5, 0),
        complex(0, 4)
        ]
```

```python
second_vk = first_vk.copy()

AKF = []

for i in range(len(first_vk)):

    sum_1 = 0
    sum_2 = 0
    product = 0

    for i in range(len(first_vk)):
        elem_1 = first_vk[i]
        elem_2 = second_vk[i]

        product += (elem_1 * elem_2.conjugate())

        sum_1 += pow((abs(elem_1)), 2)
        sum_2 += pow((abs(elem_2)), 2)

    AKF.append(product / (pow(sum_1, 0.5) * pow(sum_2, 0.5)))

    element = second_vk.pop(0)
    second_vk.append(element)

print("АКФ фигуры:")
print(AKF)
```

```python
haar = get_veivlet(len(first_vk))

print("Вейвлет:")
print(haar)
```

```python
print("АКФ со свёрткой:")
print(convolve_vk(AKF, haar))
```
