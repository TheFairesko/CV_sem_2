## 1

```python
first_vk = [
        complex(5, 0),
        complex(0, -4),
        complex(-5, 0),
        complex(0, 4)
        ]

second_vk = [
        complex(-5, 0),
        complex(0, 4),
        complex(5, 0),
        complex(0, -4)
        ]
```

```python
sum_1 = 0
sum_2 = 0
product = 0

for i in range(len(first_vk)):
    elem_1 = first_vk[i]
    elem_2 = second_vk[i]

    product += (elem_1 * elem_2.conjugate())

    sum_1 += pow((abs(elem_1)), 2)
    sum_2 += pow((abs(elem_2)), 2)

result = product / (pow(sum_1, 0.5) * pow(sum_2, 0.5))

print(result)
```
