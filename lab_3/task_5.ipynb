```python
import numpy as np
from PIL import Image
```

```python
image_name = "cloud"

image_path = f"origins/{image_name}.jpg"
image_path_save_1 = f"results/task_5/{image_name}_roberts.png"
image_path_save_2 = f"results/task_5/{image_name}_prewitt.png"
image_path_save_3 = f"results/task_5/{image_name}_sobel.png"
```

## Робертс

```python
def roberts_operator(image_array: np.ndarray) -> np.ndarray:
    """Применяет оператор Робертса для выделения контуров на изображении"""
    # Ядра Робертса для горизонтального и вертикального градиентов
    kernel_x = np.array([[1, 0],
                         [0, -1]])

    kernel_y = np.array([[0, 1],
                         [-1, 0]])

    # Получаем размеры изображения
    height, width, _ = image_array.shape

    # Создаем пустой массив для хранения градиентов
    gradient_x = np.zeros_like(image_array, dtype=np.float32)
    gradient_y = np.zeros_like(image_array, dtype=np.float32)

    # Применяем ядра Робертса
    for i in range(height - 1):
        for j in range(width - 1):
            for k in range(3): 
                gradient_x[i, j, k] = np.sum(image_array[i:i+2, j:j+2, k] * kernel_x)
                gradient_y[i, j, k] = np.sum(image_array[i:i+2, j:j+2, k] * kernel_y)

    # Вычисляем общий градиент и нормализуем градиент изображения
    gradient_magnitude = np.sqrt(gradient_x**2 + gradient_y**2)
    gradient_magnitude = (gradient_magnitude / gradient_magnitude.max()) * 255

    # Преобразуем в целочисленный тип
    gradient_magnitude = gradient_magnitude.astype(np.uint8)

    return gradient_magnitude
```

```python
image = Image.open(image_path).convert("RGB")
rgb_array = np.array(image)

roberts_image = roberts_operator(rgb_array)

Image.fromarray(roberts_image).save(image_path_save_1)
```

## Превитт

```python
def prewitt_operator(image_array: np.ndarray) -> np.ndarray:
    """Применяет оператор Превитта для выделения контуров на изображении"""
    # Ядра Прюитта для горизонтального и вертикального градиентов
    kernel_x = np.array([[-1, 0, 1],
                         [-1, 0, 1],
                         [-1, 0, 1]])

    kernel_y = np.array([[-1, -1, -1],
                         [0,  0,  0],
                         [1,  1,  1]])

    # Получаем размеры изображения
    height, width, _ = image_array.shape

    # Создаем пустой массив для хранения градиентов
    gradient_x = np.zeros_like(image_array, dtype=np.float32)
    gradient_y = np.zeros_like(image_array, dtype=np.float32)

    # Применяем ядра Прюитта
    for i in range(1, height - 1):
        for j in range(1, width - 1):
            for k in range(3): 
                gradient_x[i, j, k] = np.sum(image_array[i-1:i+2, j-1:j+2, k] * kernel_x)
                gradient_y[i, j, k] = np.sum(image_array[i-1:i+2, j-1:j+2, k] * kernel_y)

    # Вычисляем общий градиент и нормализуем градиент изображения
    gradient_magnitude = np.sqrt(gradient_x**2 + gradient_y**2)

    gradient_magnitude = (gradient_magnitude / gradient_magnitude.max()) * 255

    # Преобразуем в целочисленный тип
    gradient_magnitude = gradient_magnitude.astype(np.uint8)

    return gradient_magnitude
```

```python
image = Image.open(image_path).convert("RGB")
rgb_array = np.array(image)

prewitt_image = prewitt_operator(rgb_array)

Image.fromarray(prewitt_image).save(image_path_save_2)
```

## Собель

```python
import numpy as np
from PIL import Image

def sobel_operator(image_array: np.ndarray) -> np.ndarray:
    """Применяет оператор Собеля для выделения контуров на изображении"""
    # Ядра Собеля для горизонтального и вертикального градиентов
    kernel_x = np.array([[-1, 0, 1],
                         [-2, 0, 2],
                         [-1, 0, 1]])

    kernel_y = np.array([[-1, -2, -1],
                         [0,  0,  0],
                         [1,  2,  1]])

    # Получаем размеры изображения
    height, width, _ = image_array.shape

    # Создаем пустой массив для хранения градиентов
    gradient_x = np.zeros_like(image_array, dtype=np.float32)
    gradient_y = np.zeros_like(image_array, dtype=np.float32)

    # Применяем ядра Собеля
    for i in range(1, height - 1):
        for j in range(1, width - 1):
            for k in range(3):  # Для каждого канала (R, G, B)
                gradient_x[i, j, k] = np.sum(image_array[i-1:i+2, j-1:j+2, k] * kernel_x)
                gradient_y[i, j, k] = np.sum(image_array[i-1:i+2, j-1:j+2, k] * kernel_y)

    # Вычисляем общий градиент и нормализуем градиент изображения
    gradient_magnitude = np.sqrt(gradient_x**2 + gradient_y**2)
    gradient_magnitude = (gradient_magnitude / gradient_magnitude.max()) * 255

    # Преобразуем в целочисленный тип
    gradient_magnitude = gradient_magnitude.astype(np.uint8)

    return gradient_magnitude
```

```python
image = Image.open(image_path).convert("RGB")
rgb_array = np.array(image)

sobel_image = sobel_operator(rgb_array)

Image.fromarray(sobel_image).save(image_path_save_3)
```
