```python
import numpy as np
from PIL import Image
```

```python
image_name = "moving_castle"

image_path = f"origins/{image_name}.jpg"
image_path_save = f"results/task_4/{image_name}_diz.png"
```

```python
def floyd_steinberg_dithering(gray_array: np.ndarray) -> np.ndarray:
    """Применяет дизеринг к изображению методом Флойда-Стейнберга"""
    # Получаем размеры изображения
    height, width = gray_array.shape

    # Создаём новый целевой массив
    output_array = gray_array.astype(np.int32) 
    
    for y in range(height):
        for x in range(width):
            # Вычисляем ошибку
            old_pixel = output_array[y, x]

            if old_pixel > 127:
                new_pixel = 255 
            else:
                new_pixel = 0

            output_array[y, x] = new_pixel
            quant_error = old_pixel - new_pixel
            
            # Распределение ошибки на соседние пиксели
            if x < width - 1:
                output_array[y, x + 1] = np.clip(output_array[y, x + 1] + quant_error * 7 // 16, 0, 255)
            if y < height - 1:
                output_array[y + 1, x] = np.clip(output_array[y + 1, x] + quant_error * 5 // 16, 0, 255)
                if x > 0:
                    output_array[y + 1, x - 1] = np.clip(output_array[y + 1, x - 1] + quant_error * 3 // 16, 0, 255)
                if x < width - 1:
                    output_array[y + 1, x + 1] = np.clip(output_array[y + 1, x + 1] + quant_error * 1 // 16, 0, 255)
    
    # Возвращаем массив к типу uint8
    return output_array.astype(np.uint8)
```

```python
image = Image.open(image_path).convert("L")
rgb_array = np.array(image)

dithered_image = floyd_steinberg_dithering(rgb_array)

Image.fromarray(dithered_image).save(image_path_save)
```
