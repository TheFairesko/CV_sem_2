```python
import numpy as np
import math

def rotate_image(rgb_array: np.ndarray, angle: float) -> np.ndarray:
    """Вращение изображения методом Македона-Оуэна."""
    # Если угол 180 градусов, то просто переворачиваем изображение
    if angle == 180:
        return np.flip(np.flip(rgb_array, axis=0), axis=1)
    
    # Преобразуем угол в радианы
    alpha = math.radians(angle)
    tan_half_alpha = math.tan(alpha / 2)
    sin_alpha = math.sin(alpha)
    cos_alpha = math.cos(alpha)
    
    # Получаем размеры изображения
    height, width, channels = rgb_array.shape
    
    # Вычисляем новые размеры изображения
    # Новые размеры должны быть достаточны для размещения повернутого изображения
    # diagonal = int(math.ceil(math.sqrt(height**2 + width**2)))
    # new_height = diagonal
    # new_width = diagonal

    # Вычисляем максимальную проекцию
    max_projection = max(height, width)
    
    # Новые размеры холста
    new_size = math.ceil(max_projection)
    new_height = height + new_size
    new_width = width + new_size
    
    # Создаем новое изображение с паддингом
    padded_array = np.zeros((new_height, new_width, channels), dtype=np.uint8)
    
    # Вычисляем смещение для центрирования исходного изображения
    y_offset = (new_height - height) // 2
    x_offset = (new_width - width) // 2
    
    # Помещаем исходное изображение в центр нового
    padded_array[y_offset:y_offset+height, x_offset:x_offset+width, :] = rgb_array
    
    # Создаем пустой массив для повернутого изображения
    rotated_array = np.zeros_like(padded_array)
    
    # Центр нового изображения
    center_x, center_y = new_width // 2, new_height // 2
    
    # Первое преобразование: сдвиг по y
    intermediate_array1 = np.zeros_like(padded_array)
    for y in range(new_height):
        for x in range(new_width):
            x_shifted = x - center_x
            y_shifted = y - center_y
            
            # Первый сдвиг (по y)
            x1 = x_shifted - y_shifted * tan_half_alpha
            y1 = y_shifted
            
            # Возвращаем координаты в исходную систему
            x1 += center_x
            y1 += center_y
            
            # Проверяем, находятся ли координаты в пределах изображения
            if 0 <= x1 < new_width and 0 <= y1 < new_height:
                intermediate_array1[y, x] = padded_array[int(y1), int(x1)]
    
    # Второе преобразование: сдвиг по x
    intermediate_array2 = np.zeros_like(intermediate_array1)
    for y in range(new_height):
        for x in range(new_width):
            x_shifted = x - center_x
            y_shifted = y - center_y
            
            # Второй сдвиг (по x)
            x2 = x_shifted
            y2 = y_shifted + x_shifted * sin_alpha
            
            # Возвращаем координаты в исходную систему
            x2 += center_x
            y2 += center_y
            
            # Проверяем, находятся ли координаты в пределах изображения
            if 0 <= x2 < new_width and 0 <= y2 < new_height:
                intermediate_array2[y, x] = intermediate_array1[int(y2), int(x2)]
    
    # Третье преобразование: сдвиг по y
    rotated_array = np.zeros_like(intermediate_array2)
    for y in range(new_height):
        for x in range(new_width):
            x_shifted = x - center_x
            y_shifted = y - center_y
            
            # Третий сдвиг (по y)
            x3 = x_shifted - y_shifted * tan_half_alpha
            y3 = y_shifted
            
            # Возвращаем координаты в исходную систему
            x3 += center_x
            y3 += center_y
            
            # Проверяем, находятся ли координаты в пределах изображения
            if 0 <= x3 < new_width and 0 <= y3 < new_height:
                rotated_array[y, x] = intermediate_array2[int(y3), int(x3)]
    
    return rotated_array
```

```python
image = Image.open(image_path).convert("RGB")
rgb_array = np.array(image, dtype = np.float32)

rotated_array = rotate_image(rgb_array, 150)

Image.fromarray(rotated_array.astype(np.uint8)).save(image_path_save)
```
