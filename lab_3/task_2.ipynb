```python
import numpy as np
from PIL import Image
```

```python
image_name = "mononoke"

image_path = f"origins/{image_name}.jpg"
image_path_save_1 = f"results/task_2/{image_name}_lin.png"
image_path_save_2 = f"results/task_2/{image_name}_nonlin.png"
```

## Линейный фильтр

```python
def gaussian_kernel(size: int, sigma: float = 1) -> np.ndarray:
    """Создает гауссово ядро заданного размера."""
    size = int(size) // 2
    x, y = np.mgrid[-size:size+1, -size:size+1]
    normal = 1 / (2.0 * np.pi * sigma**2)
    kernel = np.exp(-((x**2 + y**2) / (2.0 * sigma**2))) * normal
    return kernel / np.sum(kernel) 
```

```python
def apply_gaussian_blur(image: np.ndarray, kernel_size: int = 7, sigma: float = 1) -> np.ndarray:
    """Применяет гауссово размытие к изображению."""
    kernel = gaussian_kernel(kernel_size, sigma)
    pad_size = kernel_size // 2
    
    padded_image = np.pad(image, ((pad_size, pad_size), (pad_size, pad_size), (0, 0)), mode='linear_ramp')
    output = np.zeros_like(image)

    for c in range(image.shape[2]):
        for i in range(image.shape[0]):
            for j in range(image.shape[1]):
                output[i, j, c] = np.sum(padded_image[i:i+kernel_size, j:j+kernel_size, c] * kernel)
    
    return output
```

```python
image = Image.open(image_path).convert("RGB")
rgb_array = np.array(image)

filtered_array = apply_gaussian_blur(rgb_array)

Image.fromarray(filtered_array.astype(np.uint8)).save(image_path_save_1)
```

## Нелинейный фильтр

```python
def apply_median_filter(image: np.ndarray, kernel_size: int = 7) -> np.ndarray:
    """Применяет медианный фильтр к изображению."""
    pad_size = kernel_size // 2

    padded_image = np.pad(image, ((pad_size, pad_size), (pad_size, pad_size), (0, 0)), mode='linear_ramp')
    output = np.zeros_like(image)

    for c in range(image.shape[2]):
        for i in range(image.shape[0]):
            for j in range(image.shape[1]):
                neighborhood = padded_image[i:i+kernel_size, j:j+kernel_size, c]
                output[i, j, c] = np.median(neighborhood)
    
    return output
```

```python
image = Image.open(image_path).convert("RGB")
rgb_array = np.array(image)

filtered_array = apply_median_filter(rgb_array)

Image.fromarray(filtered_array.astype(np.uint8)).save(image_path_save_2)
```
